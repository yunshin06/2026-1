```python
# N x N 행렬의 모든 값을 0보다 크고 (N x N x 10) 보다 작은 임의의(random) 값으로 채워주는 함수
import random
N = int(input())
def random_matrix(n):
  matrix = []
  for i in range(n):
    row = []
    for j in range(n):
      row.append(random.randrange(1, (n*n*10)))
    matrix.append(row)
  return matrix
```

```python
# 행렬 N x N을 예쁘게 출력하는 함수
def print_matrix(matrix):
  n = len(matrix)
  for i in range(n):
    for j in range(n):
      print(f'{matrix[i][j]:5d}', end=' ')
    print()
```

```python
# N x N 행렬 세개 A, B, C를 곱하고 더하는, 즉, A x B + C 하는 프로그램
a =  random_matrix(N)
b =  random_matrix(N)
c =  random_matrix(N)
ans = []
for i in range(N):
  row = []
  for j in range(N):
    row.append(a[i][j]*b[i][j]+c[i][j])
  ans.append(row)
print_matrix(ans)
```
