```python
import os

dir1 = input("첫 번째 디렉토리 이름을 입력하세요: ").strip()
dir2 = input("두 번째 디렉토리 이름을 입력하세요: ").strip()

def get_file_info(path, base_path):

    file_dict = {}
    try:
        with os.scandir(path) as entries:
            for entry in entries:
                if entry.is_dir():
                    file_dict.update(get_file_info(entry.path, base_path))
                elif entry.is_file():
                    rel_path = os.path.relpath(entry.path, base_path)

                    file_size = entry.stat().st_size

                    try:
                        with open(entry.path, 'rb') as f:
                            file_content = f.read()
                    except Exception:
                        file_content = b""

                    file_dict[rel_path] = (file_size, file_content)
    except FileNotFoundError:
        print(f"오류: '{path}' 경로를 찾을 수 없음.")
    return file_dict

info1 = get_file_info(dir1, dir1)
info2 = get_file_info(dir2, dir2)

print("\n=== 비교 결과 ===")

if len(info1) != len(info2):
    print(f" 불일치: 파일 개수가 다릅니다. ({dir1}: {len(info1)}개, {dir2}: {len(info2)}개)")
else:
    if info1 == info2:
        print(f"일치: 두 디렉토리의 파일 개수({len(info1)}개), 이름, 크기, 내용이 모두 동일")
    else:
        print("불일치: 파일 개수는 같으나 이름, 크기, 또는 내용이 다른 파일이 존재")
```
